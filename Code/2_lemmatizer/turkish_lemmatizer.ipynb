# Turkish Lemmatizer
This lemmatizer was built for TCLex. The lemmatizer is publicly available at [HuggingFace](https://huggingface.co/umit/turkish-lemmatizer). See model card for details.

This Python Notebook illustrates how to use it.

#1. Install Dependencies

```python
# Install spacy-transformers,
# A runtime restart might be required after installation
!pip install -U spacy spacy-transformers huggingface_hub --quiet
```

```python
#import spacy and huggingface
import spacy
from huggingface_hub import snapshot_download

# download the model
model_path = snapshot_download("umit/turkish-lemmatizer")

# load spaCy pipeline
nlp = spacy.load(model_path)
```

#2. Sample Usage

```python
#lemmatize text

text = "Ayşe kitapları masanın üzerine koydu."

doc = nlp(text)

lemmas = [token.lemma_ for token in doc]

print(lemmas)
```
